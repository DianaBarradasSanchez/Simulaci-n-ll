```python
import numpy as np

np.random.seed(42)
T = 1.0
N = 1000        # Pasos de tiempo (dt = 0.001)
M = 100000      # Número de trayectorias
dt = T / N

# Simulación de incrementos brownianos dW ~ N(0, dt)
dW = np.sqrt(dt) * np.random.randn(M, N)

# Construcción de trayectorias W_s con W_0 = 0
W = np.hstack([np.zeros((M, 1)), np.cumsum(dW, axis=1)])

# Aproximación de la integral \int_0^1 W_s^2 ds por suma de Riemann (extremo izquierdo)
integrales = np.sum(W[:, :-1]**2 * dt, axis=1)

# Estimadores
media_sim = np.mean(integrales)
var_sim = np.var(integrales, ddof=1)

print(f"Esperanza teórica: 0.5000 | Simulada: {media_sim:.4f}")
print(f"Varianza teórica:  0.3333 | Simulada: {var_sim:.4f}")
```
